# LA TABLE

Сегодня на семинаре мы поговорим с вами о захватывающей новинке в мире машинного обучения - модели LaTable.

Представьте себе: табличные данные повсюду вокруг нас, от финансовых отчетов до медицинских карт. Но вот незадача - создавать их искусственно оказалось куда сложнее, чем генерировать тексты или изображения. Почему? Да потому что каждая таблица уникальна: разные типы данных, загадочные заголовки столбцов, отсутствие четкой структуры.

И тут на сцену выходит наш герой - **LaTable**. Эта модель не боится разнообразия таблиц. Она учится на самых разных данных и создает новые, поразительно похожие на настоящие. А если ее немного подучить, она справляется даже с совершенно незнакомыми типами таблиц.

Конечно, у LaTable есть свои слабости. Например, она пока не умеет работать "с нуля", без предварительного обучения. Но именно изучение таких ограничений поможет нам в будущем создать еще более совершенные модели для работы с табличными данными.




И так мы поговорим о создании фундаментальных моделей для табличных данных. Давайте рассмотрим, почему это так важно и интересно.

## Мотивация

Представьте себе, что в мире текстов и изображений уже существуют мощные модели, способные творить чудеса. Они могут генерировать тексты, создавать изображения, и даже решать сложные задачи без специального обучения. Но в мире табличных данных такого прорыва пока не произошло. И это странно, ведь табличные данные окружают нас повсюду:

- В медицине это истории болезней и результаты анализов
- В финансах - отчеты и прогнозы
- В науке - результаты экспериментов
- В государственном управлении - данные переписи населения

Создание мощной модели для работы с табличными данными могло бы произвести революцию во всех этих областях!

## Почему это сложно?

Разработка такой модели - непростая задача. Вот с какими трудностями сталкиваются исследователи:

1. Разнообразие данных: каждая таблица уникальна, с собственным набором признаков.
2. Отсутствие структуры: в отличие от текста или изображений, в таблицах нет четкого порядка столбцов.
3. Смешанные типы данных: в одной таблице могут сочетаться числа, категории и пропущенные значения.

## Что мы хотим получить?

Наша цель - создать модель, которая сможет:

1. Работать с разными наборами данных, генерируя таблицы с различным количеством и типами признаков.
2. Одинаково хорошо обрабатывать как числовые, так и категориальные данные.
3. Использовать текстовый контекст - описания данных, названия признаков и т.д.
4. Не зависеть от порядка столбцов в таблице.

Такая модель открыла бы новые горизонты в анализе данных, позволяя генерировать синтетические данные, улучшать существующие наборы и даже предсказывать недостающие значения.

В мире машинного обучения для табличных данных существует несколько подходов. Давайте рассмотрим их:

## 1. Модели для отдельных наборов данных

- Существуют специальные модели для табличных данных, включая диффузионные модели.
- Однако большинство из них плохо работают с категориальными данными и не могут генерировать данные для разных таблиц.

## 2. Подходы на основе языковых моделей

- Используют предобученные языковые модели (например, GPT-2) для работы с таблицами.
- Преимущества: простота, возможность работать с сырыми данными.
- Недостатки:
  - Очень медленные и дорогие в обучении и использовании.
  - Неэффективны для числовых данных, так как моделируют их как текст.
  - Плохо справляются с аппроксимацией непрерывных распределений.

## 3. LaTable - новый подход

- Использует сквозное числовое представление данных и фреймворк диффузионной модели.
- Преодолевает ограничения языковых моделей.
- Может работать с разными типами данных и разными таблицами.





## METHOD

**LaTable**

1. **Основа модели**:
   - Использует трансформер (только энкодер) для работы с разными типами данных.
   - Применяет диффузионный подход, который постепенно улучшает качество генерируемых данных.

2. **Входные данные**:
   - Зашумленные значения признаков (это часть диффузионного процесса).
   - Метаданные: описание набора данных, названия столбцов.
   - Дополнительная информация: маска условий и маска пропусков.

3. **Обработка входных данных**:
   - Текстовые данные (метаданные и категории) преобразуются в числовые векторы с помощью предобученной языковой модели.
   - Все входные данные приводятся к единому формату с помощью специальных нейронных сетей (MLP).
   - Полученные данные объединяются и подаются на вход трансформеру.

4. **Выход модели**:
   - Трансформер обрабатывает данные.
   - Результат декодируется отдельно для каждого признака.
   - Для категориальных признаков модель выдает вероятности для каждой категории.

LaTable - это шаг вперед в работе с табличными данными, так как она может обрабатывать разные типы данных, учитывать контекст (метаданные) и генерировать реалистичные синтетические данные.

Представьте, что LaTable - это умная фабрика по производству табличных данных. Вот как она работает:

1. **Гибкость входных данных**:
   Эта фабрика может принимать любые таблицы - большие и маленькие, с разным количеством столбцов, с числами или текстом. Она не привередлива!

2. **Предобработка данных**:
   - Для текстовых данных у нас есть специальный переводчик (LLM энкодер). Он превращает слова в числа, которые машина понимает лучше. Этот переводчик уже обучен и не меняется.
   - Числовые данные проходят через "шумилку" - специальный процесс, который немного искажает данные. Это помогает модели лучше учиться.

3. **Главный механизм - трансформер**:
   Это сердце нашей фабрики. Он обрабатывает все данные, но не заботится о том, в каком порядке они поступили. Он также не пытается предсказать будущее на основе прошлого (поэтому нет "причинной маскировки").

4. **Дополнительные возможности**:
   Мы можем легко добавить extra информацию. Например, сказать модели, какие данные отсутствуют или задать особые условия для генерации.

5. **Универсальность**:
   Этот же механизм используется и для обучения модели, и для создания новых данных. Это как учиться готовить и готовить на одной и той же кухне.

### Простыми словами:

1. **Почему диффузионная модель?**
   - Диффузионные модели стали популярными, потому что они могут генерировать данные более высокого качества по сравнению с другими подходами, такими как GANs (генеративно-состязательные сети) и VAEs (вариационные автоэнкодеры).
   - Для LaTable используется конкретный вид диффузионной модели — *Denoising Diffusion Implicit Model* (DDIM), которая работает с дискретным временем и управляет процессом добавления и удаления шума.

2. **Что нужно учесть?**
   - Чтобы модель соответствовала требованиям D1-D4 (гибкость для разных таблиц, работа с числовыми и категориальными данными, использование метаданных, независимость от порядка столбцов), важно тщательно выбрать архитектуру.
   - Необходимо учитывать, как обрабатывать одновременно числовые и категориальные признаки, а также как интегрировать метаданные таблиц в процесс генерации.

3. **Что дальше?**
   - Авторы обещают в следующих разделах подробно объяснить, как каждый элемент LaTable помогает удовлетворить эти требования.

### Основная идея:
Диффузионная модель выбрана за ее высокое качество генерации данных. Однако для работы с табличными данными требуется особый подход к архитектуре, чтобы учитывать специфику таблиц: смешанные типы данных, метаданные и гибкость структуры.



Авторы также объясняют, как они используют трансформер в качестве основы модели LaTable для удовлетворения двух ключевых требований:

1. **D1: Работа с разными наборами данных**
   - Используется трансформер только с энкодером.
   - Он может обрабатывать входные и выходные данные переменной длины.
   - Это позволяет обучать модель на разных наборах данных с различным количеством признаков.

2. **D4: Независимость от порядка столбцов**
   - В модели не используется позиционное кодирование.
   - Это делает трансформер инвариантным к порядку входных данных.
   - Другими словами, порядок столбцов в таблице не влияет на результат.

Авторы отмечают, что рекуррентная архитектура не смогла бы обеспечить такую независимость от порядка.

Также упоминается, что скрытое измерение трансформера обозначается как dh.

Таким образом, выбор трансформера в качестве основы позволяет LaTable работать с разнообразными табличными данными, независимо от их структуры и порядка столбцов.


**ИСПОЛЬЗОВАНИЕ КОНТЕКСТА**

В этом пункте 3.3.2 авторы объясняют, как LaTable использует контекст (D3), чтобы лучше понимать данные. Вот основные моменты:

1. **Что такое контекст?**
   - Контекст — это текстовая информация, такая как описание набора данных, названия столбцов и категории в таблице.
   - Например, если в таблице есть столбцы "gender" и "sex", модель должна понимать, что это похожие понятия.

2. **Как используется контекст?**
   - Для обработки текстовых данных (контекста) используется предварительно обученная языковая модель (LLM), которая превращает текст в числовые векторы (embedding).
   - Это позволяет не обучать модель с нуля для понимания текстов и использовать сходства между словами.

3. **Как это работает?**
   - Описание набора данных кодируется в вектор.
   - Категориальные переменные (например, "gender: male, female") превращаются в строки и тоже кодируются в векторы.
   - Названия столбцов таблицы также преобразуются в числовые представления.

4. **Почему это важно?**
   - Использование LLM помогает модели лучше понимать текстовые данные даже при ограниченном количестве табличных данных.
   - Это делает модель более универсальной и способной работать с разными наборами данных.

5. **Какая LLM используется?**
   - Авторы выбрали модель WhereIsAI/UAE-Large-V1, потому что она:
     - Открыта и доступна.
     - Показывает высокие результаты на задачах текстового кодирования.
     - Легковесна (всего 1.34 ГБ).

### Простыми словами:
LaTable использует заранее обученную языковую модель для преобразования текста (описания данных, названий столбцов, категорий) в числовые представления. Это помогает модели лучше понимать контекст таблиц и справляться с задачами генерации данных даже при ограниченном количестве исходных данных.

**3.3.3 Mixed-type variables (D2)**

В этом разделе авторы описывают, как LaTable обрабатывает числовые и категориальные переменные, чтобы генерировать табличные данные.

### 1. Обработка числовых переменных:
- **Процесс диффузии**: Для числовых переменных используется стандартный процесс диффузии. На каждом временном шаге добавляется шум к значениям.
- **Преобразование**: Зашумленные значения чисел преобразуются с помощью мелкой нейронной сети (MLP) в формат, который может использовать трансформер.
- **Выход**: После обработки трансформером, выходные данные снова преобразуются в числовые значения, и вычисляется ошибка (loss) для каждого признака.

### 2. Обработка категориальных переменных:
- **Вдохновение от CDCD**: Для категориальных переменных авторы используют подход, основанный на Continuous Diffusion for Categorical Data (CDCD). Здесь каждую категорию представляют в виде высокоразмерного вектора.
- **Добавление шума**: К этим векторам добавляется шум, как и в случае с числовыми переменными.
- **Контекстуальные векторы**: Вместо того чтобы учить отдельные векторы для каждой категории, используются предварительно обученные векторы от языковой модели (LLM), которые учитывают контекст названий категорий.
- **Преобразование и вывод**: Эти зашумленные векторы также преобразуются в формат для трансформера и затем обратно в вероятности для каждой категории.

### 3. Объединение потерь:
- Ошибка для категориальных переменных вычисляется с помощью кросс-энтропии (cross-entropy), а для числовых — с помощью среднеквадратичной ошибки (MSE).
- Общая ошибка модели складывается из потерь по числовым и категориальным переменным.

### 4. Итог:
Модель LaTable использует несколько независимых мелких нейронных сетей для приведения различных типов данных (числовых и категориальных) к единому формату. Входные данные подаются в трансформер, который затем генерирует выходные данные для дальнейшей обработки.

Таким образом, LaTable эффективно обрабатывает как числовые, так и категориальные данные, обеспечивая гибкость и точность при генерации табличных данных.

Формула объединяет два типа ошибок:
- Для категорий она проверяет, насколько правильно модель предсказывает вероятности для каждой категории (кросс-энтропия).

Давайте рассмотрим простой Google Colab блокнот, который демонстрирует основные идеи из статьи о LaTable. Мы реализуем упрощенную версию модели, которая работает с табличными данными и использует трансформер для генерации данных.

Вот шаги, которые мы реализуем:

### 1. Импорт библиотек
Начнем с установки и импорта необходимых библиотек.

```python
# Установка необходимых библиотек
!pip install torch transformers numpy pandas matplotlib scikit-learn

# Импорт библиотек
import torch
import torch.nn as nn
import torch.optim as optim
from transformers import AutoModel, AutoTokenizer
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
import matplotlib.pyplot as plt
```



### 2. Создание синтетического набора данных
Создаем простой табличный набор данных с числовыми и категориальными признаками.



```python
# Функция для создания синтетических данных
def create_synthetic_data(num_samples=1000):
    data = {
        'age': np.random.randint(18, 80, num_samples),
        'income': np.random.randint(20000, 100000, num_samples),
        'gender': np.random.choice(['male', 'female'], num_samples),
        'purchased': np.random.choice([0, 1], num_samples)
    }
    return pd.DataFrame(data)

# Создаем данные
df = create_synthetic_data()
print("Пример данных:")
print(df.head())

# Кодируем категориальные признаки и нормализуем числовые
label_encoders = {}
scaler = MinMaxScaler()

df['gender'] = LabelEncoder().fit_transform(df['gender'])  # male -> 0, female -> 1
df[['age', 'income']] = scaler.fit_transform(df[['age', 'income']])

# Разделяем данные на обучающую и тестовую выборки
train_data, test_data = train_test_split(df, test_size=0.2, random_state=42)
```

### 3. Определение модели
Создаем упрощенную модель с трансформером для обработки табличных данных.




```python
class SimpleTabularTransformer(nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super(SimpleTabularTransformer, self).__init__()
        self.embedding = nn.Linear(input_dim, hidden_dim)  # Входной слой для преобразования в скрытое пространство
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=hidden_dim, nhead=4), num_layers=2
        )
        self.output_layer = nn.Linear(hidden_dim, input_dim)  # Выходной слой

    def forward(self, x):
        x = self.embedding(x)
        x = self.transformer(x)
        x = self.output_layer(x)
        return x

# Параметры модели
input_dim = train_data.shape[1]  # Количество признаков (столбцов)
hidden_dim = 16

# Создаем модель
model = SimpleTabularTransformer(input_dim=input_dim, hidden_dim=hidden_dim)
print(model)
```


### 4. Обучение модели
Реализуем процесс обучения с использованием MSE (среднеквадратичной ошибки) для числовых данных.



```python
# Подготовка данных для PyTorch
train_tensor = torch.tensor(train_data.values, dtype=torch.float32)
test_tensor = torch.tensor(test_data.values, dtype=torch.float32)

# Оптимизатор и функция потерь
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.MSELoss()

# Обучение модели
epochs = 100
losses = []

for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    outputs = model(train_tensor)
    loss = criterion(outputs, train_tensor)  # Сравниваем предсказания с оригинальными данными

    loss.backward()
    optimizer.step()

    losses.append(loss.item())
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")

# График ошибки обучения
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```



### 5. Генерация новых данных
Используем обученную модель для генерации синтетических данных.



```python
# Генерация новых данных (на основе тестовых данных)
model.eval()
with torch.no_grad():
    generated_data = model(test_tensor).numpy()

# Преобразуем данные обратно в DataFrame для наглядности
generated_df = pd.DataFrame(generated_data, columns=df.columns)

# Визуализируем оригинальные и сгенерированные данные (например, распределение возраста)
plt.hist(test_data['age'], bins=20, alpha=0.5, label='Original')
plt.hist(generated_df['age'], bins=20, alpha=0.5, label='Generated')
plt.legend()
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Original vs Generated Data (Age)")
plt.show()
```

Сравнение распределений оригинальных и сгенерированных данных показывает, что модель смогла уловить общую структуру данных. Однако есть несколько моментов, которые можно обсудить:
